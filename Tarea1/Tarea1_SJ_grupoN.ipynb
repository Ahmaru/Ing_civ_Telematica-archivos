#### Pregunta 4: 	¿Qué es una shell de Linux y qué implicancias tiene un permiso SETUID o SUID?  
##### R: En sistemas operativos basados en linux, una shell es el intermediario entre el usuario y el kernel.  Nosotros cuando usamos comandos en la terminal, la shell se encarga de que estos comandos se ejecuten correctamente. Los terminos SETUID y SUID estan directamente relacionados con la shell, SETUID (set user ID) es un permiso que se le da a un archivo mediante los comandos : (ejemplos usando ej 4.1)  
```bash
$ sudo chmod 4111 ~/mysh
$ sudo chmod a-rw+s ~/mysh
```
#### Y el permiso SUID es practicamente lo mismo, solamente que al hacer la llamada al sistema aparece como SUID que es forma abreviada.  


#### Pregunta 5: Proponga una línea de comando que busque todos los archivos que tengan el permiso activo SETUID o SUID.  
##### R: Usando el comando ```find ``` en nuestra terminal con el siguiente formato:  
```bash
$ find /ruta_donde_buscar -type f -perm -u=s -exec 
```
##### Lo que hace esta linea de codigo es mostrar todos los archivos que tengan el permiso de SETUID (super user) dentro de la ruta de busqueda especificada, este comando se puede variar en los tipos de permisos que tenga los archivos, por ejemplo, (```-a=x ó -g=w```), haciendo esto las busquedas que hacemos pueden ser más rapidas. 

#### Pregunta 6: ¿Qué es un proceso en segundo plano y cómo podría incidir la herramienta NOHUP en la ejecución de un proceso?  
##### Un proceso en segundo plano lo podemos definir como un proceso que esta siendo ejecutado por debajo de la interfaz que nosotros como usuarios vemos o interactuamos, esto ocurre constantemente en sistemas operativos más actuales ya que el usuario no ve los procesos que se hacen al abrir software, *por ejemplo el navegador*, bajo esta misma idea podemos decir que los procesos en segundo plano son automaticos y necesarios.  
##### Aunque nosotros como usuarios tambien podemos definir si queremos algun proceso, comando o aplicacion se ejecute en segundo plano, en sistemas linux podemos usar la terminal usando comandos como:  
```bash
$ firefox & 
$ jupyter-notebook & 
```   
##### Como ya definimos procesos en segundo plano, podemos hablar de la herramienta **NOHUP**, con los ejemplos de arriba si uno cierra la terminal donde se esta ejecutando en segundo plano este proceso o aplicacion terminara abruptamente su ejecucion, para evitar esto existe la herramienta **NOHUP**.  
```bash
$ nohup virt-manager &
```  
##### Esto hace que aunque se cierre la terminal donde se ejecuto el software *virt-manager*, este seguira ejecutandose en segundo plano
