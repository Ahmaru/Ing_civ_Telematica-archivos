### Preguntas 7-8
7.
```c
#include <sys/types.h> // Esta librería provee del tipo de dato pid_t
#include <stdio.h>     // Standard Input/Output
#include <unistd.h>    // Esta librería facilita la función fork()
#include <stdlib.h>    // Entrega facilidades para usar el dato NULL
#include <sys/wait.h>  // Esta librería proporciona la función wait()

int value = 5; // se designa un valor
int main()     // el programa comienza
{
    pid_t pid; // se inicializa una variable pid_t llamada pid
    pid = fork(); // la funciíon fork() crea un subproceso y le da el valor del id de proceso a la variable pid
    if (pid == 0) {  // el subproceso es el que tiene este valor
            value += 15; // este agrega 15 a la variable value
            return 0; // el subproceso termina
    }
    else if (pid > 0) { // el proceso principal posee un valor de id de proceso no negativo 
            wait(NULL); // la función wait() hace que este proceso principal espere a los subprocesos a terminar antes de seguir
            printf("PARENT: value = %d\n",value); /* LINE A */ // función que imprime el valor final de value
            return 0; // se termina el programa
    }
}
```

8. La función fork creará un subproceso. Este proceso y el proceso principal pasan por todo el código hacia adelante por su propia cuenta. Dado que tienen pid distintos, estos tendrán efectos distintos y realizarán acciones distintas.
