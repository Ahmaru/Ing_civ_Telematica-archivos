#### Pregunta 6: ¿Qué es un proceso en segundo plano y cómo podría incidir la herramienta NOHUP en la ejecución de un proceso?  
##### Un proceso en segundo plano lo podemos definir como un proceso que esta siendo ejecutado por debajo de la interfaz que nosotros como usuarios vemos o interactuamos, esto ocurre constantemente en sistemas operativos más actuales ya que el usuario no ve los procesos que se hacen al abrir software, *por ejemplo el navegador*, bajo esta misma idea podemos decir que los procesos en segundo plano son automaticos y necesarios.  
##### Aunque nosotros como usuarios tambien podemos definir si queremos algun proceso, comando o aplicacion se ejecute en segundo plano, en sistemas linux podemos usar la terminal usando comandos como:  
```bash
$ firefox & 
$ jupyter-notebook & 
```   
##### Como ya definimos procesos en segundo plano, podemos hablar de la herramienta **NOHUP**, con los ejemplos de arriba si uno cierra la terminal donde se esta ejecutando en segundo plano este proceso o aplicacion terminara abruptamente su ejecucion, para evitar esto existe la herramienta **NOHUP**.  
```bash
$ nohup virt-manager &
#Ejemplo con software de virtualizacion virt-manager en linux.
```  
##### Esto hace que aunque se cierre la terminal donde se ejecuto el software *virt-manager*, este seguira ejecutandose en segundo plano

### Pregunta 7: Explique línea a línea el propósito del código MySample.
#### 
```c
#include <sys/types.h> // Esta librería provee del tipo de dato pid_t
#include <stdio.h>     // Standard Input/Output
#include <unistd.h>    // Esta librería facilita la función fork()
#include <stdlib.h>    // Entrega facilidades para usar el dato NULL
#include <sys/wait.h>  // Esta librería proporciona la función wait()

int value = 5; // se designa un valor
int main()     // el programa comienza
{
    pid_t pid; // se inicializa una variable pid_t llamada pid
    pid = fork(); // la funciíon fork() crea un subproceso y le da el valor del id de proceso a la variable pid
    if (pid == 0) {  // el subproceso es el que tiene este valor
            value += 15; // este agrega 15 a la variable value
            return 0; // el subproceso termina
    }
    else if (pid > 0) { // el proceso principal posee un valor de id de proceso no negativo 
            wait(NULL); // la función wait() hace que este proceso principal espere a los subprocesos a terminar antes de seguir
            printf("PARENT: value = %d\n",value); /* LINE A */ // función que imprime el valor final de value
            return 0; // se termina el programa
    }
}
```

### Pregunta 8: ¿Cuál es el objetivo de la función fork() en el código MySample?  
#### La función fork creará un subproceso. Este proceso y el proceso principal pasan por todo el código hacia adelante por su propia cuenta. Dado que tienen pid distintos, estos tendrán efectos distintos y realizarán acciones distintas.

### Pregunta 10: ¿Cuánto tiempo tarda el upgrade de Kernel y cómo puede revertirlo?  
#### El tiempo que tarda sera dependiente a la potencia del hardware de la maquina que estemos usando para hacer el upgrade de kernel, las maquinas de hoy en dia pueden hacer esta tarea bastante rapido (en aprox 5 o más minutos), aunque con maquinas de gama intermedia puede llegar a los 20 minutos aprox. Tambien puede depender de la cantidad de paquetes necesarios para la mejora.  
#### Para revertir un upgrade de kernel, lo podemos hacer mediante el *GRUB*, seleccionando las opciones avanzadas y seleccionar kernel previo. Otra opcion es cambiando la variable GRUB_DEFAULT en la carpeta /etc/default/grub, esto funciona ya teniendo diferentes kernel en el sistema (El segundo funciona en sistemas linux como Ubuntu 25.10 que usamos en el ramo).
